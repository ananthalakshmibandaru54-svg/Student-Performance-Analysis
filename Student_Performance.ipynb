```python
import pandas as pd
```

```python
data = {
    "Name": ["Anu", "Kamesh", "Pandu", "Anantha", "Gowri"],
    "Class": ["A", "A", "B", "B", "A"],
    "Python": [85, 65, 92, 30, 78],
    "SQL": [90, 70, 95, 45, 82],
    "Maths": [88, 72, 89, 50, 80]
}

df = pd.DataFrame(data)

df
```

```python
print("Total students:", len(df))
print("Column names:", df.columns.tolist())
```

```python
df["Average"] = (df["Python"] + df["SQL"] + df["Maths"]) / 3

df
```

```python
df["Average"] = df["Average"].round(2)

df
```

```python
def get_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 75:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 35:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average"].apply(get_grade)

df
```

```python
df[["Name", "Average", "Grade"]]
```

```python
df["Result"] = "Pass"

df.loc[
    (df["Python"] < 35) |
    (df["SQL"] < 35) |
    (df["Maths"] < 35),
    "Result"
] = "Fail"

df
```

```python
print("Passed:", (df["Result"] == "Pass").sum())
print("Failed:", (df["Result"] == "Fail").sum())
```

```python
topper = df.loc[df["Average"].idxmax()]

print("Topper:", topper["Name"])
print("Average Marks:", topper["Average"])
```

```python
passed = (df["Result"] == "Pass").sum()
total = len(df)

percentage = (passed / total) * 100

print("Pass Percentage:", percentage, "%")
```

```python
print("Total Students:", len(df))
print("Class Average:", round(df["Average"].mean(), 2))
print("Passed Students:", (df["Result"] == "Pass").sum())
print("Failed Students:", (df["Result"] == "Fail").sum())
print("Pass Percentage:", round(percentage, 2), "%")
print("Topper:", topper["Name"])
```

```python
import matplotlib.pyplot as plt
```

```python
plt.bar(df["Name"], df["Average"])

plt.xlabel("Student Name")
plt.ylabel("Average Marks")
plt.title("Student Performance Analysis")

plt.show()
```

```python
df.to_csv("student_report.csv", index=False)

print("Report saved successfully!")
```

```python
import sqlite3
import pandas as pd

con = sqlite3.connect("students.db")

name = input("Enter student name: ")
class_name = input("Enter class: ")
python = int(input("Enter Python marks: "))
sql = int(input("Enter SQL marks: "))
maths = int(input("Enter Maths marks: "))

average = round((python + sql + maths) / 3, 2)

if average >= 90:
    grade = "A"
elif average >= 75:
    grade = "B"
elif average >= 60:
    grade = "C"
elif average >= 35:
    grade = "D"
else:
    grade = "F"

if python < 35 or sql < 35 or maths < 35:
    result = "Fail"
else:
    result = "Pass"

new_student = pd.DataFrame([{
    "Name": name,
    "Class": class_name,
    "Python": python,
    "SQL": sql,
    "Maths": maths,
    "Average": average,
    "Grade": grade,
    "Result": result
}])

new_student.to_sql("students", con, if_exists="append", index=False)

print("Student added successfully!")
```

```python
import sqlite3
import pandas as pd

con = sqlite3.connect("students.db")

name = input("Enter student name: ")

python = int(input("Enter new Python marks: "))
sql = int(input("Enter new SQL marks: "))
maths = int(input("Enter new Maths marks: "))

average = round((python + sql + maths) / 3, 2)

if average >= 90:
    grade = "A"
elif average >= 75:
    grade = "B"
elif average >= 60:
    grade = "C"
elif average >= 35:
    grade = "D"
else:
    grade = "F"

if python < 35 or sql < 35 or maths < 35:
    result = "Fail"
else:
    result = "Pass"

query = """
UPDATE students
SET Python = ?, SQL = ?, Maths = ?,
    Average = ?, Grade = ?, Result = ?
WHERE Name = ?
"""

cursor = con.cursor()
cursor.execute(query, (python, sql, maths, average, grade, result, name))
con.commit()

if cursor.rowcount > 0:
    print("Student marks updated successfully!")
else:
    print("Student not found")

con.close()
```

```python
import sqlite3
import pandas as pd

con = sqlite3.connect("students.db")

while True:
    print("\n--- Student Performance Analysis System ---")
    print("1. Display All Students")
    print("2. Search Student")
    print("3. Show Topper")
    print("4. Show Class Statistics")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        df = pd.read_sql_query("SELECT * FROM students", con)
        display(df)

    elif choice == "2":
        name = input("Enter student name: ")
        df = pd.read_sql_query(
            "SELECT * FROM students WHERE Name = ?",
            con,
            params=(name,)
        )

        if df.empty:
            print("Student not found")
        else:
            display(df)

    elif choice == "3":
        df = pd.read_sql_query("SELECT * FROM students", con)

        if df.empty:
            print("No student records found")
        else:
            topper = df.loc[df["Average"].idxmax()]
            print("Topper:", topper["Name"])
            print("Average:", topper["Average"])

    elif choice == "4":
        df = pd.read_sql_query("SELECT * FROM students", con)

        if df.empty:
            print("No student records found")
        else:
            print("Total Students:", len(df))
            print("Class Average:", round(df["Average"].mean(), 2))
            print("Passed:", (df["Result"] == "Pass").sum())
            print("Failed:", (df["Result"] == "Fail").sum())

    elif choice == "5":
        print("Thank you!")
        break

    else:
        print("Invalid choice. Try again.")

con.close()
```
